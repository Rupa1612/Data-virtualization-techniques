```python
df = pd.read_csv("diabetes_prediction_dataset.csv")
```

```python
df.head()
df.info()
df.shape
df.columns
```

```python
df.columns = df.columns.str.strip()

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.lower()

df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna()

df.shape
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

df.shape
```

```python
df.describe()
```

```python
df.describe(include='object')
```

```python
df.to_csv("cleaned_diabetes_dataset.csv", index=False)
```

```python
df.to_json("cleaned_diabetes_dataset.json", orient="records", indent=4)
```

```python
connection = sqlite3.connect("diabetes.db")
df.to_sql("diabetes", connection, if_exists="replace", index=False)
```

```python
pd.read_sql("SELECT * FROM diabetes LIMIT 5;", connection)
```

```python
pd.read_sql(
    "SELECT * FROM diabetes WHERE diabetes = 1;",
    connection
)
```

```python
pd.read_sql(
    "SELECT * FROM diabetes ORDER BY age DESC;",
    connection
)
```

```python
pd.read_sql("""
SELECT gender, COUNT(*) AS total_people
FROM diabetes
GROUP BY gender;
""", connection)
```

```python
pd.read_sql("""
SELECT COUNT(*) AS total_records
FROM diabetes;
""", connection)
```

```python
pd.read_sql("""
SELECT AVG(HbA1c_level) AS average_hba1c
FROM diabetes;
""", connection)
```

```python
pd.read_sql("""
SELECT MAX(blood_glucose_level) AS max_glucose
FROM diabetes;
""", connection)
```

```python
pd.read_sql("""
SELECT MIN(blood_glucose_level) AS min_glucose
FROM diabetes;
""", connection)
```

```python
pd.read_sql("""
SELECT SUM(blood_glucose_level) AS total_glucose
FROM diabetes;
""", connection)
```

```python
pd.read_sql("""
SELECT gender, COUNT(*) AS total
FROM diabetes
GROUP BY gender
HAVING COUNT(*) > 100;
""", connection)
```

```python
pd.read_sql("""
SELECT *
FROM diabetes
LIMIT 10;
""", connection)
```

```python
connection.close()
```
