```python
import pandas as pd

df = pd.read_csv("diabetes_prediction_dataset.csv")

print(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="HbA1c_level",
    y="blood_glucose_level"
)

plt.title("HbA1c Level vs Blood Glucose Level")
plt.xlabel("HbA1c Level")
plt.ylabel("Blood Glucose Level")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df["blood_glucose_level"])

plt.title("Blood Glucose Level Outlier Detection")

plt.show()
```

```python
plt.figure(figsize=(10,7))

corr = df[
    [
        "age",
        "hypertension",
        "heart_disease",
        "bmi",
        "HbA1c_level",
        "blood_glucose_level",
        "diabetes"
    ]
].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm"
)

plt.title("Diabetes Dataset Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="diabetes"
)

plt.title("Diabetes Categories")
plt.xlabel("Diabetes")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.hist(
    df["age"],
    bins=30
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
diabetes = df["diabetes"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    diabetes,
    labels=["No Diabetes", "Diabetes"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Diabetes Distribution")

plt.show()
```

```python
gender = df.groupby("gender")["blood_glucose_level"].mean()

gender = gender.sort_values(
    ascending=False
)

plt.figure(figsize=(8,5))

gender.plot(kind="bar")

plt.title("Average Blood Glucose Level by Gender")

plt.ylabel("Average Blood Glucose Level")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["blood_glucose_level"],
    bins=30,
    kde=True
)

plt.title("Blood Glucose Level Distribution")

plt.show()
```

```python
sns.pairplot(
    df[
        [
            "age",
            "bmi",
            "HbA1c_level",
            "blood_glucose_level",
            "diabetes"
        ]
    ]
)

plt.show()
```

```python
plt.figure(figsize=(9,6))

plt.scatter(
    df["HbA1c_level"],
    df["blood_glucose_level"],
    s=df["bmi"] * 5,
    alpha=0.5,
    c=df["diabetes"],
    cmap="viridis"
)

plt.colorbar(label="Diabetes")

plt.xlabel("HbA1c Level")
plt.ylabel("Blood Glucose Level")

plt.title("Bubble Chart: HbA1c vs Blood Glucose")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.violinplot(
    data=df,
    x="diabetes",
    y="blood_glucose_level"
)

plt.title("Blood Glucose Distribution by Diabetes")

plt.xlabel("Diabetes")
plt.ylabel("Blood Glucose Level")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x="diabetes",
    y="blood_glucose_level"
)

plt.title("Blood Glucose Level by Diabetes Category")

plt.xlabel("Diabetes")
plt.ylabel("Blood Glucose Level")

plt.show()
```
